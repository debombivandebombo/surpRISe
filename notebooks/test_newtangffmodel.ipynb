```python
# %matplotlib widget

import numpy as np
import plotly.io as pio
from scipy.constants import c, pi

pio.renderers.default = (
    "notebook_connected"  # 'browser' for avoid error but shown in the browser
)

# import surpRISe
from surpRISe.channels.tx2ris2rx.tang_ff_channel import tang_ff_channel
from surpRISe.components.antenna import Antenna
from surpRISe.components.ris import RIS
from surpRISe.utils.utils import *
```

```python
# Create RIS
f = 28e9
lam = c / f

ris = RIS(
    center=[0, 0, 0],
    boresight="z",
    Ncol=32,
    Nrow=32,
    delta_h=lam / 2,
    delta_v=lam / 2,
    dx=0.47 * lam,
    dy=0.47 * lam,
    fc=f,
    radiation_factor=1,
    phase_shifts_dict={"0": 0, "1": pi},
    att_dict={"0": 1, "1": 1},
)
```

```python
max_bf = (ris.Nrow * ris.Ncol) ** 2

# Create antennas and channel
tx_pos = sph2cart([10, 0, 0])
tx_antenna = Antenna(tx_pos, 0)

rx_pos1 = sph2cart([10, 180, 30])
rx_antenna1 = Antenna(rx_pos1, 0)

rx_pos2 = sph2cart([10, 180, 45])
rx_antenna2 = Antenna(rx_pos2, 0)

rx_pos = np.array([[0, 0, 20], rx_pos1[0], rx_pos2[0]])

channel = tang_ff_channel()

# Generate rx locations as a 2D map
ris.reset_ris()
ris.plot_phase_dist()
print(
    channel.phase_bf_multrx(tx_antenna=tx_antenna, rx_positions=rx_pos, ris=ris)
    / max_bf
)

channel.optimize(
    optimizer="direct_disc",
    mode="cont",
    wave_model="spherical",
    tx_antenna=tx_antenna,
    rx_antenna=rx_antenna1,
    ris=ris,
)
ris.plot_phase_dist()
print(
    channel.phase_bf_multrx(tx_antenna=tx_antenna, rx_positions=rx_pos, ris=ris)
    / max_bf
)

channel.optimize(
    optimizer="direct_disc",
    mode="cont",
    wave_model="plane",
    tx_antenna=tx_antenna,
    rx_antenna=rx_antenna1,
    ris=ris,
)
ris.plot_phase_dist()
print(
    channel.phase_bf_multrx(tx_antenna=tx_antenna, rx_positions=rx_pos, ris=ris)
    / max_bf
)

channel.optimize(
    optimizer="direct_disc",
    mode="cont",
    wave_model="spherical",
    tx_antenna=tx_antenna,
    rx_antenna=rx_antenna2,
    ris=ris,
)
ris.plot_phase_dist()
print(
    channel.phase_bf_multrx(tx_antenna=tx_antenna, rx_positions=rx_pos, ris=ris)
    / max_bf
)

channel.optimize(
    optimizer="direct_disc",
    mode="cont",
    wave_model="plane",
    tx_antenna=tx_antenna,
    rx_antenna=rx_antenna2,
    ris=ris,
)
ris.plot_phase_dist()
print(
    channel.phase_bf_multrx(tx_antenna=tx_antenna, rx_positions=rx_pos, ris=ris)
    / max_bf
)
```
