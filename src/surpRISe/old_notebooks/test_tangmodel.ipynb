```python
%matplotlib widget

import matplotlib.pyplot as plt
import numpy as np
import plotly.io as pio
from scipy.constants import c

pio.renderers.default = (
    "notebook_connected"  # 'browser' for avoid error but shown in the browser
)

from channels.tx_ris_rx_channel import tx_ris_rx_channel
from components.antenna import Antenna
from components.ris import RIS
from utils.utils import *
```

```python
# Create RIS
f = 27e9
lam = c / f

mmwave_ris1 = RIS(
    center=[0, 0, 0],
    boresight="z",
    Ncol=56,
    Nrow=20,
    delta_h=1.4e-3,
    delta_v=2.8e-3,
    dx=1.4e-3,
    dy=2.8e-3,
    fc=f,
    radiation_factor=1,
    phase_shifts_dict={"0": np.deg2rad(165), "1": 0},
    att_dict={"0": 0.9, "1": 0.7},
    name="mmwave RIS1",
)
mmwave_ris1.plot_phase_dist()

f = 33e9
lam = c / f

mmwave_ris2 = RIS(
    center=[0, 0, 0],
    boresight="z",
    Ncol=40,
    Nrow=40,
    delta_h=3.8e-3,
    delta_v=3.8e-3,
    dx=3.8e-3,
    dy=3.8e-3,
    fc=f,
    radiation_factor=1,
    phase_shifts_dict={"0": np.deg2rad(150), "1": 0},
    att_dict={"0": 0.8, "1": 0.8},
    name="mmwave RIS2",
)
mmwave_ris2.plot_phase_dist()
```

```python
# Antenna params
# G_dBi_27ghz = 20.4
# G_dBi_33ghz = 21.1

# Adding the losses of their RF cables (Gline)
G_dBi_27ghz_A = (
    2.9 / 2 + 10
)  # Gt + Gr + Gline = 2.9 dB in setup A, compensating the Ptx it is + 20 dBm
G_dBi_27ghz_B = (
    24 / 2 + 10
)  # Gt + Gr + Gline = 24 dB in setup B, compensating the Ptx it is + 20 dBm
G_dBi_33ghz_B = (
    22 / 2 + 10
)  # Gt + Gr + Gline = 22 dB in setup B, compensating the Ptx it is + 20 dBm
```

```python
# Fig. 9:
# mmwave_RIS1, f = 27e9, P_tx = 20 dBm, system A
# d_tx = 1.3 m, theta_tx = 0, phi_tx = 0
# d_rx = 2.6 m, theta_rx in [0, 90], phi_tx = [0, 180] --> all the angles in the horizontal plane

# Fixed position for TX:
tx_pos = sph2cart([1.3, 0, 0])
tx_antenna = Antenna(position=tx_pos, gain_dbi=G_dBi_27ghz_A)
tx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

# Auxiliar fixed position for RX (RIS not optimized so not needed):
rx_pos = sph2cart([2.6, 0, 0])
rx_antenna = Antenna(position=rx_pos, gain_dbi=G_dBi_27ghz_A)
rx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

# Create channel
channel = tx_ris_rx_channel(
    tx_antenna=tx_antenna, rx_antenna=rx_antenna, ris=mmwave_ris1, model="Tang_general"
)

# With default state (all "0" states) obtain the RX power at all the positions of the RX
channel.plot_path_gain_angle(r=2.6, Nr=200, axis="x", vmin=-105, vmax=-55)
```

```python
# Fig. 10:
# mmwave_RIS1, f = 27e9, P_tx = 20 dBm, system B
# d_tx = d_rx = 2 m, theta_tx = theta_rx in [5, 50] deg, phi_tx = 0 deg, phi_rx = 180 deg

ii = 0
pg = np.zeros((10, 1))

for kk in range(5, 51, 5):
    # Fixed position for TX:
    tx_pos = sph2cart([2, 0, kk])
    tx_antenna = Antenna(position=tx_pos, gain_dbi=G_dBi_27ghz_B)
    tx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

    # Fixed position for RX:
    rx_pos = sph2cart([2, 180, kk])
    rx_antenna = Antenna(position=rx_pos, gain_dbi=G_dBi_27ghz_B)
    rx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

    # Create channel
    channel = tx_ris_rx_channel(
        tx_antenna=tx_antenna,
        rx_antenna=rx_antenna,
        ris=mmwave_ris1,
        model="Tang_general",
    )

    # Obtain the RX power at the position of the RX
    pg[ii] = 10 * np.log10(channel.path_gain_multrx(rx_pos))
    ii += 1

pg_plot = pg[:, 0] - np.max(pg[:, 0])

plt.figure(figsize=(6, 5))
plt.plot(
    np.arange(5, 51, 5),
    pg_plot,
    marker="o",
    color="blue",
    label="mmwave RIS1, cos(theta)",
)
plt.xlabel("Theta RX (deg)")
plt.ylabel("Path Gain normalized(dB)")
plt.title("Fig. 10")
plt.ylim(-5, 0)
plt.xlim(5, 50)
plt.legend()
```

```python
# Fig. 12a:
# mmwave_RIS1, f = 27e9, P_tx = 20 dBm, intelligent reflection, system A
# d_tx = 1.3 m, theta_tx = 0, phi_tx = 0
# d_rx = 2.6 m, thata_rx in [0, 90], phi_rx = [0, 180] --> all the angles in the horizontal plane

# Intelligent refletion here is obtained by a stripe code where (m in [1, 56]):
# -     code state "0" is applied when mod(m,14) in [1,7]
# -     code state "1" is applied when mod(m,14) not in [1,7]

# Create TX and RX
tx_pos = sph2cart([1.3, 0, 0])
tx_antenna = Antenna(position=tx_pos, gain_dbi=G_dBi_27ghz_A)
tx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

# Auxiliar fixed position for RX (RIS not optimized because of it so not needed):
rx_pos = sph2cart([2.6, 0, 0])
rx_antenna = Antenna(position=rx_pos, gain_dbi=G_dBi_27ghz_A)
rx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

# Give the stripe code to the RIS and create the channel
stripe_code = np.zeros((mmwave_ris1.Nrow, mmwave_ris1.Ncol))
stripe_code[:, 7:14] = 1  # Apply state "1" in columns 7 to 13
stripe_code[:, 21:28] = 1  # Apply state "1" in columns 21 to 27
stripe_code[:, 35:42] = 1  # Apply state "1" in columns 35 to 41
stripe_code[:, 49:56] = 1  # Apply state "1" in columns 49 to 55

mmwave_ris1.set_all_state(stripe_code)

mmwave_ris1.plot_phase_dist()

# Create channel
channel = tx_ris_rx_channel(
    tx_antenna=tx_antenna, rx_antenna=rx_antenna, ris=mmwave_ris1, model="Tang_general"
)
# With actual state (stripe code) obtain the RX power at all the positions of the RX
channel.plot_path_gain_angle(r=2.6, Nr=200, axis="x", vmin=-105, vmax=-55)


# Fig. 12b:
# mmwave_RIS1, f = 27e9, P_tx = 20 dBm, intelligent reflection, system B
# d_tx = 0.5 m, theta_tx = 0, phi_tx = 0
# d_rx in [0.5, 5] m, theta_rx = 34 deg, phi_rx = 180 deg

# Create TX and RX
tx_pos = sph2cart([0.5, 0, 0])
tx_antenna = Antenna(position=tx_pos, gain_dbi=G_dBi_27ghz_B)
tx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

# Auxiliar fixed position for RX (RIS not optimized because of it so not needed):
rx_pos = sph2cart([1, 180, 34])
rx_antenna = Antenna(position=rx_pos, gain_dbi=G_dBi_27ghz_B)
rx_antenna.point_to(mmwave_ris1.center)  # Pointing towards the RIS

# Create channel
channel = tx_ris_rx_channel(
    tx_antenna=tx_antenna, rx_antenna=rx_antenna, ris=mmwave_ris1, model="Tang_general"
)

# Plot the path gain distribution for the given parameters
channel.plot_path_gain_dist(
    phi_deg=180, theta_deg=34, r1=0.3, r2=5, Nr=100, vmin=-45, vmax=-10
)


# Reset RIS to default state
mmwave_ris1.reset_ris()
```
