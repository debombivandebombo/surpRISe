## What to study
Near-field (spherical wave) and Far-field (plane wave) phase estimation on RIS with continuous (perfect) and discrete (k-bit) phase shifts
## Steps:
1. Create the RIS --> Greenerwave FR2 RIS model is used to later measure in real scenario which has 1-bit configurability
2. Create channel model --> Tang general model (`tang_general_channel()`) is used to be as realistic as possible (spherical wave assumption)
3. Create TX and RX parameters --> gain (dBi), radiation pattern (everything except the antenna position). Parameters are based on mmWave antennas that will be used in real measurements
4. Create positions (angles and distances) for both TX and RX for all the configurations to be studied --> (next points)
5. TX at theta = [0, 15, 30, 45, 60] deg, phi = 0, and distances of [very close to RIS (1-2 m), before the ff dist of discrete shifts for that TX angle (-1 m of the border), after the ff dist of disc. shifts for that TX angle (+1-2 m of the border), before the normal ff dist for that TX angle (-1 m of the border), after the normal ff dist for that TX angle (+1-2 m of the border), further from the normal ff dist for that TX angle (much more distance of the border)].
6. RX at theta = [0, 15, 30, 45, 60] deg, phi = 180, and distances from 1 m to further from the normal ff dist for that TX angle (much more distance of the border) in steps of 0.5-1 m
7. For each configuration of TX fixed position, all the RX positions should be tried (all angles and distances)
8. Calculations should be of the `tang_general_channel()` with phase estimations with both 'spherical' and 'plane' wave model and with both 'continuous' and 'discrete' phase shifts of the RIS
9. First trails can check the viability of each of the discrete aproximations for RIS phase shifts with the actual 3 optimizers (direct_disc, max_bf, and min_phase_error (Algorithm 1 of the paper)) --> check the speed and performance of each one to do the rest of the simulations of the study with the best optimizer
10. Extra cases of TX and RX positions with other phi != 0, 180 to represent real scenario elevations (even if they will not be real measured)
